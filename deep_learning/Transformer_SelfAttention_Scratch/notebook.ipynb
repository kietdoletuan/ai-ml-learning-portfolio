```python
import numpy as np

np.random.seed(42)
seq_len = 3
d_k = 4

Q = np.random.randn(seq_len, d_k)
K = np.random.randn(seq_len, d_k)
V = np.random.randn(seq_len, d_k)
```

```python
def compute_scores(Q, K, V, d_k):
    scores = (Q @ K.T)/(np.sqrt(d_k))
    
    return scores

def softmax(X):
    return np.exp(X)/np.sum(np.exp(X), axis=1, keepdims=True)
```

```python
def self_attention(Q, K, V, d_m):
    scores = compute_scores(Q, K, V, d_m)
    print(f"Scores shape: {scores.shape}")

    weights = softmax(scores)
    print(f"Weights shape: {weights.shape}")
    print(weights.sum(axis=1))
    
    output = weights @ V
    print(f"Output shape: {output.shape}")

    return output
```

```python
output = self_attention(Q, K, V, d_k)
print(output)
```

```python
import torch

Q_tensor = torch.Tensor(Q)
K_tensor = torch.Tensor(K)
V_tensor = torch.Tensor(V)

torch_output = torch.nn.functional.scaled_dot_product_attention(Q_tensor, K_tensor, V_tensor)
print(torch_output, torch_output.shape)

```
