```python
from anomalib.data import MVTec
from anomalib.models import Padim
from anomalib.engine import Engine
import matplotlib.pyplot as plt
import numpy as np
import torch
```

```python
data_module = MVTec(
    root="C:/mvtec_ad",
    category="bottle",
    train_batch_size=32,
    eval_batch_size=32,
    num_workers=0
)

model = Padim(
    backbone="resnet18",
    layers=["layer1", "layer2", "layer3"]
)

engine = Engine(
    max_epochs = 1,
    image_metrics=["AUROC"],
    pixel_metrics=["AUROC"],
    accelerator = "auto"
)
```

```python
engine.fit(model=model,datamodule=data_module)
results = engine.test(model=model,datamodule=data_module)

print(results)
```

```python
predictions = engine.predict(model=model, datamodule=data_module)
```

```python
print(type(predictions))
print(type(predictions[0]))
print(predictions[0].keys())
print(predictions[0]['anomaly_maps'].shape)
print(predictions[0]['image'].shape)
```

```python
mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
std  = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

def show_predictions(n=5):
    fig, axes = plt.subplots(3, n, figsize=(15, 3))

    defect_images = []
    defect_maps = []
    defect_masks = []

    for batch in predictions:
        labels = batch['label']
        for i in range(len(labels)):
            if labels[i] == 1:
                defect_images.append(batch['image'][i])
                defect_maps.append(batch['anomaly_maps'][i].squeeze())
                defect_masks.append(batch['mask'][i])
            if len(defect_images) == n:
                break
        if len(defect_images) == n:
            break

    for i in range(n):
        img = defect_images[i] * std + mean
        img = img.permute(1, 2, 0).numpy()
        img = img.clip(0, 1)
        axes[0, i].imshow(img)
        axes[0, i].set_title("Image Input")
        axes[0, i].axis("off")

        axes[1, i].imshow(defect_masks[i])
        axes[1, i].set_title("Masks")
        axes[1, i].axis("off")

        axes[2, i].imshow(defect_maps[i], cmap="hot")
        axes[2, i].set_title("Anomaly Map")
        axes[2, i].axis("off")

    plt.tight_layout()
    plt.show()
         
```

```python
show_predictions(n=5)
```

```python
import mlflow
import os

categories = ["bottle", "cable", "capsule", "carpet", "grid", "hazelnut",
              "leather", "metal_nut", "pill", "screw", "tile",
              "toothbrush", "transistor", "wood", "zipper"]

results = {}

mlflow.set_experiment("PADIM_MVTEC")
mlflow.set_tracking_uri(f"sqlite:///{os.path.abspath('../../mlflow.db')}")


for cat in categories:
    with mlflow.start_run(run_name=f"padim-{cat}"):
        print(f"==={cat}===")
        data_module = MVTec(
            root="C:/mvtec_ad",
            category=cat,
            train_batch_size=32,
            eval_batch_size=32,
            num_workers=0
        )
        model = Padim()
        engine = Engine(
            default_root_dir="C:/madet_results",
            max_epochs=1,
            image_metrics=["AUROC"],
            pixel_metrics=["AUROC"],
            accelerator="auto"
        )
        engine.fit(model=model, datamodule=data_module)
        test_results = engine.test(model=model, datamodule=data_module)
        results[cat] = test_results[0]
        mlflow.log_metric("image_AUROC", test_results[0]['image_AUROC'])
        mlflow.log_metric("pixel_AUROC", test_results[0]['pixel_AUROC'])

```
